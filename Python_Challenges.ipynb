```python
#  Challenge: Student Marks Analyzer

# Create a Python program that allows a user to input student names along with their marks and then calculates useful statistics.
# Your program should:
# 1. Let the user input multiple students with their marks (name + integer score).
# 2. After input is complete, display:
#    - Average marks
#    - Highest marks and student(s) who scored it
#    - Lowest marks and student(s) who scored it
#    - Total number of students

# Bonus:
# - Allow the user to enter all data first, then view the report
# - Format output clearly in a report-style layout
# - Prevent duplicate student names

def student_marks_analyzer():
    students = {}

    print("=== Student Marks Analyzer ===")

    # Input phase
    while True:
        name = input("Enter student name (or type 'done' to finish): ").strip()

        if name.lower() == 'done':
            break

        # Prevent duplicate names
        if name in students:
            print("⚠️ Student already exists! Try a different name.")
            continue

        try:
            marks = int(input(f"Enter marks for {name}: "))
            students[name] = marks
        except ValueError:
            print("❌ Invalid input! Marks should be an integer.")

    # If no data entered
    if not students:
        print("No student data entered.")
        return

    # Processing phase
    marks_list = list(students.values())
    avg_marks = sum(marks_list) / len(marks_list)

    highest = max(marks_list)
    lowest = min(marks_list)

    highest_students = [name for name, m in students.items() if m == highest]
    lowest_students = [name for name, m in students.items() if m == lowest]

    # Report phase
    print("\n" + "="*30)
    print("📊 STUDENT MARKS REPORT")
    print("="*30)

    print(f"Total Students: {len(students)}")
    print(f"Average Marks: {avg_marks:.2f}")

    print(f"\nHighest Marks: {highest}")
    print("Topper(s):", ", ".join(highest_students))

    print(f"\nLowest Marks: {lowest}")
    print("Needs Improvement:", ", ".join(lowest_students))

    print("="*30)


# Run the program
student_marks_analyzer()
```

```python
#  Challenge: CLI Contact Book (CSV-Powered)

# Create a terminal-based contact book tool that stores and manages contacts using a CSV file.

# Your program should:
# 1. Ask the user to choose one of the following options:
#    - Add a new contact
#    - View all contacts
#    - Search for a contact by name
#    - Exit
# 2. Store contacts in a file called `contacts.csv` with columns:
#    - Name
#    - Phone
#    - Email
# 3. If the file doesn't exist, create it automatically.
# 4. Keep the interface clean and clear.

# Example:
# Add Contact
# View All Contacts
# Search Contact
# Exit

# Bonus:
# - Format the contact list in a table-like view
# - Allow partial match search
# - Prevent duplicate names from being added


import csv
import os

FILE_NAME = "contacts.csv"


# Ensure file exists
def initialize_file():
    if not os.path.exists(FILE_NAME):
        with open(FILE_NAME, "w", newline="") as file:
            writer = csv.writer(file)
            writer.writerow(["Name", "Phone", "Email"])


# Add contact
def add_contact():
    name = input("Enter Name: ").strip()

    # Prevent duplicate names
    with open(FILE_NAME, "r") as file:
        reader = csv.reader(file)
        next(reader)
        for row in reader:
            if row[0].lower() == name.lower():
                print("⚠️ Contact already exists!")
                return

    phone = input("Enter Phone: ").strip()
    email = input("Enter Email: ").strip()

    with open(FILE_NAME, "a", newline="") as file:
        writer = csv.writer(file)
        writer.writerow([name, phone, email])

    print("✅ Contact added successfully!")


# View contacts
def view_contacts():
    with open(FILE_NAME, "r") as file:
        reader = csv.reader(file)
        data = list(reader)

        if len(data) <= 1:
            print("No contacts found.")
            return

        print("\n📒 Contact List")
        print("-" * 40)
        print(f"{'Name':<15}{'Phone':<12}{'Email'}")
        print("-" * 40)

        for row in data[1:]:
            print(f"{row[0]:<15}{row[1]:<12}{row[2]}")

        print("-" * 40)


# Search contact (partial match)
def search_contact():
    search_name = input("Enter name to search: ").lower()

    found = False

    with open(FILE_NAME, "r") as file:
        reader = csv.reader(file)
        next(reader)

        for row in reader:
            if search_name in row[0].lower():
                if not found:
                    print("\n🔍 Search Results")
                    print("-" * 40)
                    print(f"{'Name':<15}{'Phone':<12}{'Email'}")
                    print("-" * 40)

                print(f"{row[0]:<15}{row[1]:<12}{row[2]}")
                found = True

    if not found:
        print("❌ No matching contact found.")
    else:
        print("-" * 40)


# Main menu
def main():
    initialize_file()

    while True:
        print("\n=== 📞 Contact Book ===")
        print("1. Add Contact")
        print("2. View All Contacts")
        print("3. Search Contact")
        print("4. Exit")

        choice = input("Choose an option No.: ")

        if choice == "1":
            add_contact()
        elif choice == "2":
            view_contacts()
        elif choice == "3":
            search_contact()
        elif choice == "4":
            print("👋 Exiting... Goodbye!")
            break
        else:
            print("❌ Invalid choice. Try again.")


# Run program
main()
```
