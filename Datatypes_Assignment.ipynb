```python
# 4. Running Sum on list Write a program to print a list after performing running sum on it.

# 	i.e:

# 	Input:

list1 = [1,2,3,4,5,6]

# 	Output:
# 	[1,3,6,10,15,21]

running_sum = []
total = 0

for i in list1:
    total += i
    running_sum.append(total)

print(running_sum)
```

```python
# 5. You are given a list of integers. You are asked to make a list by running through elements of the list by adding all elements greater and itself.

# 	i.e. Say given list is [2,4,6,10,1] resultant list will be [22,20,10,23].

# 	For 1st element 2 ->> these are greater (4+6+10) values and 2 itself so on adding becomes 22.

# 	For 2nd element 4 ->> greater elements are (6, 10) and 4 itself, so on adding 20

# 	like wise for all other elememts.

# 	[2,4,6,10,1]-->[22,20,16,10,23]

list1 = [2,4,6,10,1]

result = []

for i in list1:
    total = 0
    for j in list1:
        if j >= i:   # include itself and greater elements
            total += j
    result.append(total)

print(result)
```

```python
# 7. Sort a list of alphanumeric strings based on product value of numeric character in it. If in any string there is no numeric character take it's product value as 1.

# 	Input:

the_list = ['1ac21', '23fg', '456', '098d','1','kls']

# 	Output:
# 	['456', '23fg', '1ac21', '1', 'kls', '098d']

the_list = ['1ac21', '23fg', '456', '098d','1','kls']

def product_of_digits(s):
    product = 1
    found_digit = False

    for ch in s:
        if ch.isdigit():
            product *= int(ch)
            found_digit = True

    return product if found_digit else 1

result = sorted(the_list, key=product_of_digits, reverse=True)

print(result)
```

```python
# 16. Write a Python function to check whether a number is perfect or not.¶
# 	A Perfect number is a number that is half the sum of all of its positive divisors (including itself).

# 	Example :

# 	The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
# 	Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.
#   The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.

def is_perfect(n):
    total = 0

    for i in range(1, n):
        if n % i == 0:
            total += i

    return total == n


# Test
num = 6

if is_perfect(num):
    print("Perfect Number")
else:
    print("Not Perfect Number")
```

```python
# 18. Write a python function that receives a list of integers and prints out a histogram of bin size 10

# 	Input:
the_list = [13,42,15,37,22,39,41,50]

# 	Output:
# 	{11-20:2,21-30:1,31-40:2,41-50:3}

def histogram(lst):
  bins = {}

  for num in lst:
    start = (num // 10) * 10 + 1
    end = start + 9

    key = f"{start}-{end}"

    bins[key] = bins.get(key, 0) + 1

  return bins

print(histogram(the_list))

```
